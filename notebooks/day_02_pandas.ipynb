```python
import pandas as pd

members = pd.DataFrame(
    {
        "member_id": [101, 102, 103, 104, 105, 106, 107, 108],
        "name": ["Rahul", "Arjun", "Priya", "Sneha", "Amit", "Neha", "Rohan", "Ananya"],
        "branch": [
            "Kolkata",
            "Delhi",
            "Kolkata",
            "Mumbai",
            "Delhi",
            "Mumbai",
            "Kolkata",
            "Delhi",
        ],
        "membership": [
            "Premium",
            "Basic",
            "Premium",
            "Basic",
            "Premium",
            "Basic",
            "Basic",
            "Premium",
        ],
    }
)

payments = pd.DataFrame(
    {
        "payment_id": [501, 502, 503, 504, 505, 506, 507, 508, 509],
        "member_id": [101, 101, 102, 103, 104, 104, 105, 107, 109],
        "amount": [1500, 500, 1000, 1800, 2000, 500, 1500, 1000, 1200],
        "payment_status": [
            "Paid",
            "Paid",
            "Paid",
            "Paid",
            "Paid",
            "Pending",
            "Paid",
            "Paid",
            "Paid",
        ],
    }
)

attendance = pd.DataFrame(
    {
        "attendance_id": [801, 802, 803, 804, 805, 806, 807, 808, 809, 810, 811, 812],
        "member_id": [101, 101, 101, 102, 102, 103, 103, 104, 104, 104, 106, 108],
        "visit_date": [
            "2026-11-01",
            "2026-11-02",
            "2026-11-03",
            "2026-11-01",
            "2026-11-04",
            "2026-11-02",
            "2026-11-05",
            "2026-11-01",
            "2026-11-03",
            "2026-11-06",
            "2026-11-04",
            "2026-11-05",
        ],
    }
)


payments_members = pd.merge(members, payments, on="member_id", how="inner")
payments_members = payments_members[payments_members["payment_status"] == "Paid"]
payments_members = (
    payments_members.groupby("member_id")["amount"].sum().reset_index(name="total_paid")
)


visit_member = pd.merge(members, attendance, on="member_id", how="left")
visit_member = (
    visit_member.groupby("member_id")["visit_date"]
    .count()
    .reset_index(name="attendance_count")
)


success_payment_member = pd.merge(members, payments, on="member_id", how="inner")
success_payment_member = success_payment_member[
    success_payment_member["payment_status"] == "Paid"
]
success_payment_member = (
    success_payment_member.groupby("member_id")["payment_status"]
    .count()
    .reset_index(name="payment_count")
)


final_report = pd.merge(
    members,
    pd.merge(payments_members, visit_member, on="member_id", how="outer").fillna(0),
    on="member_id",
    how="outer",
)
final_report = final_report.merge(
    success_payment_member, on="member_id", how="outer"
).fillna(0)
final_report.to_json("../datasets/final_report.json", indent=2, orient="records")


highest_revenue = (
    final_report.groupby("branch")[["total_paid"]].sum()["total_paid"].idxmax()
)
membership_type_highest_revenue = (
    final_report.groupby("membership")[["total_paid"]].sum()["total_paid"].idxmax()
)
members_zero_visits = (
    final_report[final_report["attendance_count"] == 0]["name"]
    .reset_index(drop=True)
    .to_numpy()
)
pending_payments = payments[payments["payment_status"] == "Pending"]
members_pending_payments = pd.merge(
    members, pending_payments, on="member_id", how="inner"
)["name"].to_numpy()
average_revenue = final_report["total_paid"].mean()

total_revenue = final_report["total_paid"].sum()
member_highest_success_transaction = final_report.loc[
    final_report["payment_count"].idxmax(), ["name", "member_id"]
]
percentage_zero_visits = (members_zero_visits.size / members["member_id"].count()) * 100


# First five records
print(final_report.head())

# Dataset dimensions
print(final_report.shape)

# Column information and data types
print(final_report.info())

# Statistical summary
print(final_report.describe())

# Missing values per column
print(final_report.isna().sum())
print(final_report["total_paid"].median())
print(final_report["total_paid"].mean())
```
