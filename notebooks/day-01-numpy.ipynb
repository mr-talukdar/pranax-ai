```python
import numpy as np

attendance=np.array([105,67,89,34,78])
print(attendance)
attendanceDouble=attendance*2
print(attendanceDouble)
attendanceSum=attendance.sum()
attendanceAvg=attendanceSum/attendance.size
print(attendanceSum)
print(attendanceAvg)
```

```python
import numpy as np

attendance = np.array([
    [105, 120, 98],
    [67,  89,  110],
    [78,  95,  130],
    [90,  115, 140],
    [85,  100, 125]
])

branchA_attendance=attendance[:,0]
print(branchA_attendance)
total_attendance=attendance.sum(axis=1)
print(total_attendance)
avg_attendance_each_branch=attendance.sum(axis=0)/attendance.shape[0]
print(avg_attendance_each_branch)
highest_daily_attendance=attendance.sort(0);
print(highest_daily_attendance)


```
