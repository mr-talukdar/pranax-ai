```python
import numpy as np

attendance = np.array([
    [105, 120, 98],
    [67,  89,  110],
    [78,  95,  130],
    [90,  115, 140],
    [85,  100, 125],
   
])

branchA_attendance=attendance[:,0]
print(branchA_attendance)
total_attendance=attendance.sum(axis=1)
print(total_attendance)
avg_attendance_each_branch=attendance.sum(axis=0)/attendance.shape[0]
print(avg_attendance_each_branch)
highest_daily_attendance=np.sort(attendance, axis=1)[:,(attendance.shape[1]-1)]
print(highest_daily_attendance)

total_attendance_allbranches=np.sum(attendance,axis=0)

print("The Highest attendance is of", np.argmax(total_attendance_allbranches)+1, "with", np.max(total_attendance_allbranches))



```

```python
attendance = np.array([40, 42, 43, 45, 47, 50, 120])
mean=attendance.mean()
median=np.median(attendance)
std=np.std(attendance)
var=attendance.var()
percentile=np.percentile(attendance,75)

print("Mean:", mean)
print("Median:", median)
print("Standard deviation:", std)
print("Variance:", var)
print("75th percentile:", percentile)
```

```python
gym_a = np.array([40, 45, 50])
gym_b = np.array([20, 45, 70])

gym_a_std=np.std(gym_a)
gym_b_std=np.std(gym_b)

print(gym_a_std, gym_b_std)


cv_a = (np.std(gym_a) / np.mean(gym_a)) * 100
cv_b = (np.std(gym_b) / np.mean(gym_b)) * 100

print(cv_a, cv_b)
```
