```python
import matplotlib.pyplot as plt
import pandas as pd


final_report = pd.read_json("../datasets/final_report.json")


q1 = final_report["total_paid"].quantile(0.25)
q3 = final_report["total_paid"].quantile(0.75)
iqr = q3 - q1
lower_boundary = q1 - (1.5 * iqr)
upper_boundary = q3 + (1.5 * iqr)

upper_outlier = final_report[final_report["total_paid"] > upper_boundary]
lower_outlier = final_report[final_report["total_paid"] < lower_boundary]
correlation_matrix = final_report[
    ["total_paid", "attendance_count", "payment_count"]
].corr()
correlation_matrix = final_report[
    ["total_paid", "attendance_count", "payment_count"]
].corr()

print(correlation_matrix)
attendance_fees_corr = final_report["attendance_count"].corr(final_report["total_paid"])

median_total_paid = final_report["total_paid"].median()

plt.hist(final_report["total_paid"], bins=5, edgecolor="black")


plt.title("Revenue Distribution per Member")
plt.xlabel("Total Revenue")
plt.ylabel("Number of Members")

plt.show()

plt.figure(figsize=(8, 5))

plt.scatter(final_report["attendance_count"], final_report["total_paid"])

plt.title("Attendance vs Revenue")
plt.xlabel("Attendance Count")
plt.ylabel("Total Paid (₹)")

plt.show()


plt.figure(figsize=(8, 6))

plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)


plt.colorbar()
plt.xticks(
    range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=45
)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

plt.title("PranaX Feature Correlation Matrix")
plt.tight_layout()
plt.show()

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower boundary:", lower_boundary)
print("Upper boundary:", upper_boundary)

print("Upper outliers:\n", upper_outlier)
print("Lower outliers:\n", lower_outlier)

print(attendance_fees_corr)
```
